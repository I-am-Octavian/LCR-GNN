```python
import json
import csv
import os
```

```python

# Define the directory path
directory_path = '../DATASET/Relations-fact'

for filename in os.listdir(directory_path):
    print(filename)
```

```python

# Iterate over each JSON file in the directory
for filename in os.listdir(directory_path):
    if filename.endswith(".json"):
        # Open the JSON file
        with open(os.path.join(directory_path, filename), 'r') as json_file:
            # Load the JSON data
            data = json.load(json_file)

            # Create a CSV filename based on the JSON filename
            csv_filename = filename.replace('.json', '.csv')

            # Open the CSV file for writing
            with open(os.path.join(directory_path, csv_filename), 'w', newline='') as csv_file:
                # Create a CSV writer
                csv_writer = csv.writer(csv_file)

                # Write the header row
                csv_writer.writerow(['Type','Entity1','Relationship','Type','Entity2'])

                # Iterate over each annotation in the JSON data
                for annotation in data['annotations']:
                    # Iterate over each result in the annotation
                    for result in annotation['result']:
                        # Extract the text and labels
                        text = result['value']['text']
                        labels = ', '.join(result['value']['labels'])

                        # Write the text and labels to the CSV file
                        csv_writer.writerow(['None',labels, 'is' ,'None', text])
```

```python
import os

# Define the directories
dir1 = '../DATASET/Relations-fact1/result'
dir2 = '../DATASET/Relations-issue1/result'

# Get lists of files in both directories
files_dir1 = set(os.listdir(dir1))
files_dir2 = set(os.listdir(dir2))

# Find common files
common_files = files_dir1.intersection(files_dir2)

```

```python
print(len(files_dir1))
print(len(common_files))
```

```python
# Keep only common files in dir2
for file in files_dir2:
    if file not in common_files:
        print(f'Deleting {file}')
        os.remove(os.path.join(dir2, file))
```
